```python
import pandas as pd
import numpy as np

# Criando uma tabela de pedidos do zero
dados = {
    "produto":   ["Notebook", "Camiseta", "Headphone", "Tenis", "Smartphone", "Livro", "Tablet"],
    "categoria": ["Eletronico", "Roupa", "Eletronico", "Roupa", "Eletronico", "Livro", "Eletronico"],
    "valor":     [3200.0, 89.90, 450.0, 320.0, 2800.0, 79.90, 1200.0],
    "entregue":  [True, True, False, True, False, True, True],
    "avaliacao": [5, 4, None, 5, None, 4, 5]
}

df = pd.DataFrame(dados)
print(df)
    
```

```python
# As primeiras funções que você usa em TODO projeto
print("Shape:", df.shape)           # linhas x colunas
print("\nTipos de dados:")
print(df.dtypes)
print("\nEstatísticas:")
print(df.describe())
print("\nValores nulos:")
print(df.isnull().sum())
```

```python
# Selecionando uma coluna
print(df["produto"])

# Selecionando múltiplas colunas
print(df[["produto", "valor"]])

# Filtrando linhas — só eletrônicos
eletronicos = df[df["categoria"] == "Eletronico"]
print("\nEletrônicos:")
print(eletronicos)

# Filtrando com duas condições — eletrônicos acima de R$1000
caros = df[(df["categoria"] == "Eletronico") & (df["valor"] > 1000)]
print("\nEletrônicos caros:")
print(caros)
```

```python
# Estatísticas direto na coluna
print(f"Ticket médio: R${df['valor'].mean():.2f}")
print(f"Maior valor:  R${df['valor'].max():.2f}")
print(f"Menor valor:  R${df['valor'].min():.2f}")
print(f"Total:        R${df['valor'].sum():.2f}")

# Contagem por categoria
print("\nProdutos por categoria:")
print(df["categoria"].value_counts())

# Nova coluna calculada
df["valor_com_frete"] = df["valor"] + 19.90
print("\nCom frete:")
print(df[["produto", "valor", "valor_com_frete"]])
```

```python
# Ver onde estão os nulos
print("Nulos antes:")
print(df.isnull().sum())

# Preenchendo nulos com a média
media_avaliacao = df["avaliacao"].mean()
df["avaliacao"] = df["avaliacao"].fillna(media_avaliacao)

print("\nNulos depois:")
print(df.isnull().sum())

print(f"\nMédia usada para preencher: {media_avaliacao:.1f}")
print(df[["produto", "avaliacao"]])
```

```python
# GroupBy — equivale ao GROUP BY do SQL
print("Média de valor por categoria:")
print(df.groupby("categoria")["valor"].mean().round(2))

print("\nTotal por categoria:")
print(df.groupby("categoria")["valor"].sum().round(2))

print("\nContagem e média juntos:")
print(df.groupby("categoria").agg(
    quantidade=("produto", "count"),
    ticket_medio=("valor", "mean"),
    total=("valor", "sum")
).round(2))
```

```python
import pandas as pd

musicas = pd.DataFrame({
    "artista":  ["Sorriso Maroto", "Sorriso Maroto", "Matue", "Matue", 
                 "Thiaguinho", "Thiaguinho", "WIU", "WIU"],
    "musica":   ["Nao Precisa", "Me Ve", "CANGURU", "MONSTRO",
                 "Prometo", "Veleiro", "Ta OK", "Quero Mais"],
    "genero":   ["Pagode", "Pagode", "Trap", "Trap",
                 "Pagode", "Pagode", "Trap", "Trap"],
    "streams":  [850000, 620000, 2100000, 1800000,
                 940000, 710000, 1500000, 980000],
    "likes":    [42000, 31000, 95000, 87000,
                 48000, 35000, 72000, 51000]
})

print(musicas)
```

```python
# DESAFIO 1 — músicas acima de 1 milhão
print("=== Músicas acima de 1M de streams ===")
print(musicas[musicas["streams"] > 1000000])

# DESAFIO 2 — total e média por gênero
print("\n=== Streams por gênero ===")
print(musicas.groupby("genero").agg(
    total_streams=("streams", "sum"),
    media_streams=("streams", "mean")
).round(0))

# DESAFIO 3 — taxa de engajamento
musicas["taxa_engajamento"] = (musicas["likes"] / musicas["streams"]) * 100
print("\n=== Taxa de engajamento ===")
print(musicas[["artista", "musica", "taxa_engajamento"]].round(2))

# DESAFIO 4 — artista com maior média
print("\n=== Média de streams por artista ===")
print(musicas.groupby("artista")["streams"].mean().sort_values(ascending=False).round(0))
```
