```python
frutas = ["maçã", "banana", "laranja"]

for fruta in frutas:
    print(fruta)
```

```python
linguagens = ("Python", "JavaScript", "C++")

for i, lang in enumerate(linguagens):
    print(f"Posição {i}: {lang}")
```

```python
numeros = [10, 20, 30, 40]
i = 0

while i < len(numeros):
    print(f"Índice {i} = {numeros[i]}")
    i += 1  # Incrementar é essencial para evitar loop infinito!
```

```python
aluno = {
    "nome": "Carlos",
    "curso": "Filosofia",
    "nota": 9.5
}
for chave in aluno:
    print(chave, "->", aluno[chave])  # Acessa o valor pela chave
```

```python
for valor in aluno.values():
    print(valor)
```

```python
for chave, valor in aluno.items():
    print(f"{chave.capitalize()}: {valor}")
```

```python
# Exemplo com break e continue em uma lista
numeros = [1, 2, 3, 4, 5, 6, 7, 8]

for num in numeros:
    if num % 2 == 0:
        continue  # Pula os números pares
    if num > 5:
        break     # Para o loop se o número for maior que 5
    print(num)    # Imprime apenas: 1, 3, 5
```
