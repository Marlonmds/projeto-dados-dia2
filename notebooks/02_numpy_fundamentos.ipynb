```python
import numpy as np

lista_python = [1200, 980, 1540, 720, 2100, 1800, 950]
array_numpy  = np.array([1200, 980, 1540, 720, 2100, 1800, 950])

print("Lista Python:", lista_python)
print("Array NumPy:", array_numpy)
print("Tipo:", type(array_numpy))
print("Shape:", array_numpy.shape)
print("Dtype:", array_numpy.dtype)
```

```python
import time

lista = list(range(1_000_000))
array = np.array(lista)

inicio = time.time()
total = 0
for x in lista:
    total += x
tempo_python = time.time() - inicio

inicio = time.time()
total_np = np.sum(array)
tempo_numpy = time.time() - inicio

print(f"Loop Python: {tempo_python:.4f} segundos")
print(f"NumPy:       {tempo_numpy:.4f} segundos")
print(f"NumPy foi {tempo_python/tempo_numpy:.0f}x mais rápido!")
```

```python
vendas = np.array([1200, 980, 1540, 720, 2100, 1800, 950])

print("Vendas originais:", vendas)
print("Com 10% de aumento:", vendas * 1.10)
print("Com desconto R$100:", vendas - 100)

print("\n--- Estatísticas ---")
print(f"Total:   R${np.sum(vendas):.2f}")
print(f"Média:   R${np.mean(vendas):.2f}")
print(f"Maior:   R${np.max(vendas):.2f}")
print(f"Menor:   R${np.min(vendas):.2f}")
print(f"Desvio:  R${np.std(vendas):.2f}")
```

```python
vendas = np.array([1200, 980, 1540, 720, 2100, 1800, 950])

print("Primeiro dia:", vendas[0])
print("Último dia:",   vendas[-1])
print("Dias 2 a 4:",   vendas[1:4])

meta = 1000
acima_da_meta  = vendas[vendas >= meta]
abaixo_da_meta = vendas[vendas < meta]

print(f"\nAcima da meta:  {acima_da_meta}")
print(f"Abaixo da meta: {abaixo_da_meta}")
print(f"Dias acima:  {len(acima_da_meta)}")
print(f"Dias abaixo: {len(abaixo_da_meta)}")
```

```python
lojas = np.array([
    [1200, 980,  1540, 720,  2100],
    [850,  1100, 930,  1400, 760],
    [2000, 1800, 2200, 1950, 2400]
])

print("Shape:", lojas.shape)
print("\nVendas Loja Centro:",  lojas[0])
print("Vendas Loja Barra:",   lojas[1])
print("Dia 1 todas as lojas:", lojas[:, 0])

print("\n--- Performance por loja ---")
nomes = ["Centro", "Barra", "Ipanema"]
for i, nome in enumerate(nomes):
    print(f"{nome}: média R${np.mean(lojas[i]):.2f} | total R${np.sum(lojas[i]):.2f}")

total_por_dia = np.sum(lojas, axis=0)
melhor_dia = np.argmax(total_por_dia) + 1
print(f"\nMelhor dia: Dia {melhor_dia} — R${np.max(total_por_dia):.2f}")
```

```python
# Dados reais: streams diários de uma música no Spotify
# por 2 semanas — imagine que é um pagode do Sorriso Maroto
streams = np.array([
    [45000, 52000, 48000, 61000, 89000, 120000, 95000],  # semana 1
    [78000, 82000, 71000, 88000, 105000, 143000, 112000] # semana 2
])

# DESAFIO 1: Qual o total de streams nas 2 semanas?

# DESAFIO 2: Qual a média diária de cada semana?

# DESAFIO 3: A música cresceu quantos % da semana 1 para a 2?

# DESAFIO 4: Qual foi o melhor dia (posição) e quantos streams?

# DESAFIO 5: Quantos dias passaram de 100k streams?

total = np.sum(streams)
print(f"Total de streams: {total:,}")

print(f"Média semana 1: {np.mean(streams[0]):,.0f}")
print(f"Média semana 2: {np.mean(streams[1]):,.0f}")

# Desafio 3
media_s1 = np.mean(streams[0])
media_s2 = np.mean(streams[1])
crescimento = ((media_s2 - media_s1) / media_s1) * 100
print(f"Crescimento: {crescimento:.1f}%")

# Desafio 4
melhor_streams = np.max(streams)
melhor_posicao = np.argmax(streams) + 1
print(f"Melhor dia: posição {melhor_posicao} com {melhor_streams:,} streams")

# Desafio 5
todos_os_dias = streams.flatten()
acima_100k = todos_os_dias[todos_os_dias > 100000]
print(f"Dias acima de 100k: {len(acima_100k)}")
```
