```python
vendas = [1200, 980, 1540, 720, 2100, 1800, 950]
meta = 1000

abaixo_da_meta = []
for i, venda in enumerate(vendas):
    if venda < meta:
        abaixo_da_meta.append(venda)
        print(f"Dia {i+1}:R${venda} - ABAIXO DA META !")
    else:
        print(f"Dia {i+1}: R${venda} - ok")

print(f"\nDias abaixo da meta: {len(abaixo_da_meta)}")
print(f"Total perdido: R${sum(abaixo_da_meta)}")
```

```python
pedidos = [
    {"produto": "Notebook", "categoria": "Eletronico", "valor": 3200.0, "entregue": True},
    {"produto": "Camiseta", "categoria": "Roupa", "valor": 89.90, "entregue": True},
    {"produto": "Headphone", "categoria": "Eletronico", "valor": 450.0, "entregue": False},
    {"produto": "Tenis", "categoria": "Roupa", "valor": 320.0, "entregue": True},
    {"produto": "Smartphone", "categoria": "Eletronico", "valor": 2800.0, "entregue": False},
    {"produto": "Livro", "categoria": "Livro", "valor": 79.90, "entregue": True},
    {"produto": "Tablet", "categoria": "Eletronico", "valor": 1200.0, "entregue": True},
]

total_geral = 0
for pedido in pedidos:
    total_geral = total_geral + pedido["valor"]

print(f"Total geral: R${total_geral:.2f}")
```

```python

entregues = []
total_pendente = 0
qtd_eletronicos = 0

for pedido in pedidos:
    
    # Desafio 2: produtos entregues
    if pedido["entregue"] == True:
        entregues.append(pedido["produto"])
    
    # Desafio 3: total não entregues
    if pedido["entregue"] == False:
        total_pendente = total_pendente + pedido["valor"]
    
    # Desafio 4: contar eletrônicos
    if pedido["categoria"] == "Eletronico":
        qtd_eletronicos = qtd_eletronicos + 1

print(f"Produtos entregues: {entregues}")
print(f"Valor pendente: R${total_pendente:.2f}")
print(f"Eletronicos: {qtd_eletronicos}")

```
